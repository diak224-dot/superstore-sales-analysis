# Superstore Sales Analysis

Exploratory analysis of sales, profit, customers, products, and time-based trends using the local `data/superstore.csv` dataset.

## Data Loading

```python
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
project_root = Path.cwd()
if not (project_root / "data" / "superstore.csv").is_file():
    project_root = project_root.parent

data_path = project_root / "data" / "superstore.csv"
if not data_path.is_file():
    raise FileNotFoundError(f"Dataset not found: {data_path}")

df = pd.read_csv(data_path, encoding="utf-8-sig")

visualization_dir = project_root / "visualization"
visualization_dir.mkdir(exist_ok=True)
```

## Key Performance Indicators

```python
# Calculate total sales revenue.
total_sales = df["Sales"].sum()
print("Total sales revenue:", total_sales)
```

```python
# Calculate total profit.
total_profit = df["Profit"].sum()
print("Total profit:", total_profit)
```

```python
# Calculate total units sold.
total_quantity = df["Quantity"].sum()
print("Total quantity sold:", total_quantity)
```

```python
# Calculate average sales revenue per order.
average_sales = df.groupby("Order ID")["Sales"].sum().mean()
print("Average sales per order:", average_sales)
```

```python
# Calculate average profit per record.
average_profit = df["Profit"].mean()
print("Average profit per record:", average_profit)
```

```python
# Identify the highest transaction-level sales value.
max_sales = df["Sales"].max()
print("Maximum sales value:", max_sales)
```

```python
# Identify the lowest transaction-level sales value.
min_sales = df["Sales"].min()
print("Minimum sales value:", min_sales)
```

```python
# Count unique orders.
unique_orders = df["Order ID"].nunique()
print("Unique order count:", unique_orders)
```

```python
# Count unique customers.
unique_customers = df["Customer ID"].nunique()
print("Unique customer count:", unique_customers)
```

```python
# Calculate the overall profit margin.
profit_margin = (total_profit / total_sales) * 100
print("Overall profit margin:", profit_margin, "%")
```

## Comparisons by Category, Region, and Customer Segment

```python
# Identify the category with the highest total sales revenue.
max_category_sales = df.groupby("Category")["Sales"].sum().idxmax()
print("Category with the highest sales revenue:", max_category_sales)
```

```python
# Identify the category with the highest total profit.
max_category_profit = df.groupby("Category")["Profit"].sum().idxmax()
print("Category with the highest profit:", max_category_profit)
```

```python
# Identify the sub-category with the highest total profit.
max_subcategory_profit = df.groupby("Sub-Category")["Profit"].sum().idxmax()
print("Sub-category with the highest profit:", max_subcategory_profit)
```

```python
# Rank the top 10 products by total sales revenue.
top_10_products_sales = df.groupby("Product Name")["Sales"].sum().nlargest(10)
print("Top 10 products by sales revenue:")
print(top_10_products_sales)
```

```python
# Identify the 10 products with the lowest total profit.
top_10_products_profit = df.groupby("Product Name")["Profit"].sum().nsmallest(10)
print("10 products with the lowest profit:")
print(top_10_products_profit)
```

```python
# Identify the region with the highest total sales revenue.
max_region_sales = df.groupby("Region")["Sales"].sum().idxmax()
print("Region with the highest sales revenue:", max_region_sales)
```

```python
# Identify the region with the highest total profit.
max_region_profit = df.groupby("Region")["Profit"].sum().idxmax()
print("Region with the highest profit:", max_region_profit)  
```

```python
# Identify the customer segment with the highest total sales revenue.
max_segment_sales = df.groupby("Segment")["Sales"].sum().idxmax()
print("Customer segment with the highest sales revenue:", max_segment_sales)
```

```python
# Identify the shipping mode used by the highest number of orders.
orders_by_shipping_mode = df.groupby("Ship Mode")["Order ID"].nunique()
max_shipping_mode = orders_by_shipping_mode.idxmax()
print("Most frequently used shipping mode:", max_shipping_mode)
```

```python
discount_analysis = df.groupby("Discount").agg(
    Sales=("Sales", "sum"),
    Profit=("Profit", "sum"),
)
discount_analysis["Profit Margin (%)"] = (
    discount_analysis["Profit"] / discount_analysis["Sales"] * 100
)
print("Sales and profit by discount rate:")
print(discount_analysis)
```

## Time-Series Analysis

```python
# Extract the order year from the Order Date column.
df["Order Year"] = df["Order Date"].dt.year
```

```python
# Extract the order month from the Order Date column.
df["Order Month"] = df["Order Date"].dt.month
```

```python
# Aggregate sales revenue by year.
sales_by_year = df.groupby("Order Year")["Sales"].sum()
print("Sales revenue by year:")
print(sales_by_year)
```

```python
# Aggregate sales revenue by month number.
sales_by_month = df.groupby("Order Month")["Sales"].sum()
print("Sales revenue by month:")
print(sales_by_month)
```

```python
# Identify the month with the highest sales revenue.
best_month = sales_by_month.idxmax()
print(f"Month with the highest sales revenue: {best_month}")
```

```python
# Identify the three months with the lowest sales revenue.
worst_months = sales_by_month.nsmallest(3)
print("Three months with the lowest sales revenue:")
print(worst_months)
```

```python
# Analyze annual profit trends.
profit_by_year = df.groupby("Order Year")["Profit"].sum()
print("Profit by year:")
print(profit_by_year)
```

```python
# Sales revenue by category.

# Plot sales revenue by category as a horizontal bar chart.
sales_by_category = df.groupby("Category")["Sales"].sum()
plt.figure(figsize=(10, 6))
sns.barplot(x=sales_by_category.values, y=sales_by_category.index)
plt.xlabel("Sales Revenue ($)")
plt.ylabel("Category")
plt.title("Sales Revenue by Category")
plt.tight_layout()
plt.savefig(visualization_dir / "sales_by_category.png", dpi=300, bbox_inches="tight")
plt.show()
plt.close()
```

## Sales and Profit Visualizations

```python
# Chart 2: Monthly sales revenue trend.

# Plot monthly sales revenue over time.
total_sales = df["Sales"].sum()
print("Total sales revenue:", total_sales)


monthly_sales = (
    df.set_index("Order Date")
      .resample("ME")["Sales"]
      .sum()
)

plt.figure(figsize=(14, 6))

plt.plot(
    monthly_sales.index,
    monthly_sales.values,
    marker="s",
    linewidth=2
)

plt.title("Monthly Sales Revenue Trend", fontsize=15)

plt.xlabel("Month")

plt.ylabel("Sales Revenue ($)")

plt.grid(alpha=0.3)

plt.xticks(rotation=45)

plt.tight_layout()

plt.savefig(visualization_dir / "monthly_sales_revenue.png", dpi=300, bbox_inches="tight")
plt.show()
plt.close()
```

```python
# Chart 3: Profit by region.

# Compare total profit across regions.
profit_by_region = df.groupby("Region")["Profit"].sum()
plt.figure(figsize=(5, 3))
plt.bar(profit_by_region.index, profit_by_region.values)
plt.xlabel("Region")
plt.ylabel("Profit ($)")
plt.title("Profit by Region")
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig(visualization_dir / "profit_by_region.png", dpi=300, bbox_inches="tight")
plt.show()
plt.close()
```

```python
# Chart 4: Sales revenue distribution by customer segment.

# Plot the sales revenue share by customer segment.
sales_by_segment = df.groupby("Segment")["Sales"].sum()
plt.figure(figsize=(5, 3))
plt.pie(sales_by_segment.values, labels=sales_by_segment.index, autopct="%1.1f%%")
plt.title("Sales Revenue by Customer Segment")
plt.axis("equal")
plt.savefig(visualization_dir / "sales_by_customer_segment.png", dpi=300, bbox_inches="tight")
plt.show()
plt.close()
```

```python
# Chart 5: Relationship between sales revenue and profit.

# Use a scatter plot to inspect the relationship between sales and profit.
plt.figure(figsize=(5, 3))
plt.scatter(df["Sales"], df["Profit"])
plt.xlabel("Sales Revenue ($)")
plt.ylabel("Profit ($)")
plt.title("Sales Revenue vs. Profit")
plt.tight_layout()
plt.savefig(visualization_dir / "sales_revenue_vs_profit.png", dpi=300, bbox_inches="tight")
plt.show()
plt.close()
```

```python
# Chart 6: Correlation matrix.

# Visualize pairwise correlations among numeric variables.
plt.figure(figsize=(5, 3))
correlation_matrix = df[["Sales", "Profit", "Quantity", "Discount"]].corr()
sns.heatmap(correlation_matrix, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Matrix")
plt.xlabel("Variables")
plt.ylabel("Variables")
plt.tight_layout()
plt.savefig(visualization_dir / "correlation_matrix.png", dpi=300, bbox_inches="tight")
plt.show()
plt.close()
```
